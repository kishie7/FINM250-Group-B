Bloomberg's `finance rate` is an *implied* financing rate backed out from the option mids; for a dividend payer it reflects the **net** cost of carry (financing minus the dividend), so the put-call-parity check below already accounts for NVDA's dividend without a separate term. # 1. Put-Call-Parity

Consider the option chain with expiration `2026-09-18`.

### 1.1.

For every listed strike, calculate how closely put-call parity holds. That is, compare the call spread $c-p$ to the present value moneyness, $S-K^*$.

Make a chart showing this error across strikes.

#### Note: Financing Rate

The financing rate is quoted from annual swaps with ACT/360 compounding. Thus, a quoted rate of $r$ and days-to-expiration of $\text{days}$ would be used to calculate the present value of some amount, $X$, as...

$$PV(X) = \frac{X}{1+\frac{\text{days}}{360} r}$$

```python
S = 208.64
sideways = opt.pivot_table(
    index=['days to expiration', 'strike price'],
    columns = 'option type',
    values ='price'
).reset_index().rename(columns={'Call': 'call_price', 'Put': 'put_price'})

avg_r = (opt.groupby(['strike price', 'days to expiration'])['finance rate'].mean().rename('avg_r'))

sideways = sideways.merge(avg_r, on=['strike price', 'days to expiration'])

sideways['PV_X']= sideways['strike price'] / (1 + (sideways['days to expiration'] / 360) * sideways['avg_r'])
sideways['parity_error'] = (sideways['call_price'] - sideways['put_price']) - (S - sideways['PV_X'])
plt.plot(sideways['strike price'], sideways['parity_error'])
plt.show()

sideways.head()
```

### 1.5.

Suppose that the `ATM strike = 210` trade is not an arbitrage, but rather an indication about the true financing rate.

Instead of using the given financing rate, solve for a financing rate which sets the profits (at this strike) to 0, keeping put-call parity.

Recall that we're calculating a financing rate with ACT/360 annual compounding.

```python
from scipy.optimize import fsolve

def finance_rate(true_r):
    return (20792000 * ( (1+(102/360) * true_r))) - 21000000

initial_guess = 0.04
solution = fsolve(finance_rate, initial_guess)
print(solution)
```

```python
# Parity error is very positive which means we should go for a conversion.  Short call, long put. 
shares = 100000
S = 208.64
strike_price = 270
c_p = 4.07
p_p= 50.25
r = 0.039578

# Phase 1, call short, put long, buy shares * 208.64
cash = shares * c_p
cash = cash - p_p * shares
cash = cash - (shares * S)
print(cash)

# Phase 2, hold position for 102 days
cash = cash * (1 + (102/360) * r)
print(cash)

# Phase 3, settlement date, sell shares * 270
cash = cash + (shares * 270)
print(cash)
```

### 2.1.

For the `2026-09-18` expiration, create a plot of...

```python
S = 208.64
sideways['moneyness'] = S / sideways['strike price']


plt.plot(sideways['moneyness'], sideways['call_price'], label='Call Price')
plt.plot(sideways['moneyness'], sideways['put_price'], label='Put Price')
plt.xlabel('Moneyness (S/K)')
plt.ylabel('Option Price')
plt.title('Moneyness vs. Option Value')
plt.legend()
plt.show()
```

### 2.2.

Let's now look across multiple expirations, at a single strike.

For the `ATM strike=210`, make a plot of...
* expiration date (or days to expiration) on the x-axis
* call value and put values on the y-axis

What do you notice in this relationship?

```python
expirations = ['2026-07-17', '2026-08-21', '2026-09-18', '2026-10-16', '2026-11-20']

rows = []
for exp in expirations:
    opt = pd.read_excel('option_data_bb_NVDA.xlsx', sheet_name = exp)
    strike_data = opt[opt['strike price'] == 210]
    call_row = strike_data[strike_data['option type'] == 'Call']
    put_row = strike_data[strike_data['option type'] == 'Put']
    rows.append({
        'expiration': exp,
        'days to expiration': call_row['days to expiration'].values[0],
        'call_price': call_row['price'].values[0],
        'put_price': put_row['price'].values[0],
    })


strike_210 = pd.DataFrame(rows)
print(strike_210)

plt.plot(strike_210['days to expiration'], strike_210['call_price'], label='Call')
plt.plot(strike_210['days to expiration'], strike_210['put_price'], label='Put')
plt.xlabel('Days to Expiration')
plt.ylabel('Option Price')
plt.title('Days to Expiration vs. Option Price at Strike Price $210')
plt.legend()
```
